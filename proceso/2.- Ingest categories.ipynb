```python
dbutils.widgets.removeAll()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
dbutils.widgets.text("container", "raw")
dbutils.widgets.text("catalogo", "catalog_pry")
dbutils.widgets.text("esquema", "bronze")
dbutils.widgets.text("storageName", "adlsproyectoyr1")
```

```python
container = dbutils.widgets.get("container")
catalogo = dbutils.widgets.get("catalogo")
esquema = dbutils.widgets.get("esquema")
storageName = dbutils.widgets.get("storageName")

ruta = f"abfss://{container}@{storageName}.dfs.core.windows.net/categories.csv"
```

```python
categories_schema = StructType(fields=[
                    StructField("CategoryID", IntegerType(), False),
                    StructField("CategoryName", StringType(), True),
                    StructField("ingestion_date", TimestampType(), True)
])
```

```python
categories_df = spark.read\
.option('header', True)\
.schema(categories_schema)\
.csv(ruta)
```

```python
categories_final_df = categories_df.withColumn("ingestion_date", current_timestamp())
```

```python
categories_final_df.write.mode("overwrite").insertInto(f"{catalogo}.{esquema}.categories")
```
