```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import functions as F
```

```python
dbutils.widgets.text("catalogo", "catalog_pry")
dbutils.widgets.text("esquema_source", "silver")
dbutils.widgets.text("esquema_sink", "golden")
```

```python
catalogo = dbutils.widgets.get("catalogo")
esquema_source = dbutils.widgets.get("esquema_source")
esquema_sink = dbutils.widgets.get("esquema_sink")
```

```python
df_orders_enriched_final = spark.table(f"{catalogo}.{esquema_source}.orders_enriched")
```

```python
df_sales_summary = df_orders_enriched_final.groupBy(
    col("Category"),
    col("CustomerName"),
).agg(
    sum(col("TotalValue")).alias("total_sales")
)
```

```python
df_sales_summary.withColumn("ingestion_date", current_timestamp()).write.mode("overwrite").insertInto(f"{catalogo}.{esquema_sink}.sales_summary")
```
